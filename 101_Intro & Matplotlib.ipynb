### Object Oriented Method

```python
# we start declearing or creating a figure and axes, and then by selecting the eaxes we add element onto them

fig, ax = plt.subplots(2, 1) # creates a figure and a set of subplots
ax[0].plot(age, salary, 'r')
ax[1].plot(age, salary_2, 'b')

plt.tight_layout()
```

```python
# we start declearing or creating a figure and axes, and then by selecting the eaxes we add element onto them

fig, ax = plt.subplots(1, 2) # creates a figure and a set of subplots
ax[0].plot(age, salary, 'r')
ax[1].plot(age, salary_2, 'b')

plt.tight_layout()
```

```python
# when we have multiple plots
# we start with declearing or creating a figure and axes, and then by selecting the eaxes we add element onto them

fig, ax = plt.subplots(2, 3) # creates a figure and a set of subplots
ax[0][0].plot(age, salary, 'r') # [row][column]
ax[0][1].plot(age, salary, 'r')
ax[0][2].plot(age, salary, 'r')

ax[1][0].plot(age, salary_2, 'b') # [row][column]
ax[1][1].plot(age, salary_2, 'b')
ax[1][2].plot(age, salary_2, 'b')

plt.tight_layout()
```

```python
x = np.arange(-50, 50)
y1 = np.power(x, 2)
y2 = np.power(x, 3)
y3 = np.sin(x)
y4 = np.cos(x)
y5 = np.tan(x)
y6 = np.tanh(x)
y7 = np.sinh(x)
y8 = np.exp(x)
y9 = np.cosh(x)
```

```python
fig_1, ax_1 = plt.subplots(nrows=3, ncols=3, figsize=(20,20)) # creates a figure and a set of subplots
ax_1[0, 0].plot(x, y1, 'tab:blue')
ax_1[0, 0].set_title('Square Function')
ax_1[0, 0].set_xlabel('X', fontsize=18)
ax_1[0, 0].set_ylabel('Squared Value', fontsize=18)

ax_1[0, 1].plot(x, y2, 'r')
ax_1[0, 1].set_title('Cube Function')
ax_1[0, 1].set_xlabel('X', fontsize=18)
ax_1[0, 1].set_ylabel('Cube Value', fontsize=18)

ax_1[0, 2].plot(x, y3, 'g')
ax_1[0, 2].set_title('Sin Function')
ax_1[0, 2].set_xlabel('X', fontsize=18)
ax_1[0, 2].set_ylabel('Sin Value', fontsize=18)

ax_1[1, 0].plot(x, y4, 'tab:blue')
ax_1[1, 0].set_title('cos Function')
ax_1[1, 0].set_xlabel('X', fontsize=18)
ax_1[1, 0].set_ylabel('cos Value', fontsize=18)

ax_1[1, 1].plot(x, y5, 'r')
ax_1[1, 1].set_title('tan Function')
ax_1[1, 1].set_xlabel('X', fontsize=18)
ax_1[1, 1].set_ylabel('tan Value', fontsize=18)

ax_1[1, 2].plot(x, y6, 'g')
ax_1[1, 2].set_title('tanh Function')
ax_1[1, 2].set_xlabel('X', fontsize=18)
ax_1[1, 2].set_ylabel('tanh Value', fontsize=18)

ax_1[2, 0].plot(x, y7, 'tab:blue')
ax_1[2, 0].set_title('sinh Function')
ax_1[2, 0].set_xlabel('X', fontsize=18)
ax_1[2, 0].set_ylabel('sinh Value', fontsize=18)

ax_1[2, 1].plot(x, y8, 'r')
ax_1[2, 1].set_title('exp Function')
ax_1[2, 1].set_xlabel('X', fontsize=18)
ax_1[2, 1].set_ylabel('exp Value', fontsize=18)

ax_1[2, 2].plot(x, y9, 'g')
ax_1[2, 2].set_title('cosh Function')
ax_1[2, 2].set_xlabel('X', fontsize=18)
ax_1[2, 2].set_ylabel('cosh Value', fontsize=18)

plt.tight_layout();
```
