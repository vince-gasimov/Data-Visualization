# Categorical Plots

## Count Plot

```python
sns.countplot(data = tips, x = 'day');
```

```python
sns.countplot(data = mpg, x = 'cylinders');
```

```python
fig, ax = plt.subplots()
ax = sns.countplot(data = tips, x = "day");

for p in ax.patches:
    ax.annotate((p.get_height()), (p.get_x() + 0.3, p.get_height() + 0.5))
```

- p represents a bar in a bar chart.
- .get_height() returns the height of the bar, which corresponds to the value it represents on the y-axis
- .get_x() returns the starting x-coordinate (left edge) of the bar.
- .get_height() gives the height of the bar. Adding 0.6 moves the position slightly above the bar.

```python
fig, ax = plt.subplots(figsize = (10, 5))

ax_tip = sns.countplot(data = tips, x = "day", hue = "sex")

for p in ax.patches:
    ax.annotate((p.get_height()), (p.get_x()+0.15, p.get_height()+1))
```

```python
fig, ax = plt.subplots(figsize = (10, 5))

ax_tip = sns.countplot(data = tips, x = "day", hue = "sex")

for p in ax.patches:
    ax.annotate((p.get_height()), (p.get_x()+0.15, p.get_height()+1))
    ax.text(p.get_x()+0.07, p.get_height()-4, "%"+str(round(p.get_height()/tips.day.count()*100, 2)));
```

## Barplot

- for categorical data
- comparing multiple categorical data

```python
sns.barplot(data = tips, x = 'sex', y = 'tip');

# line is confidence interval showing the uncertainty of the estimate
# by default estimator is mean
```

```python
# by default estimator is mean
sns.barplot(data = tips, x = 'sex', y = 'total_bill');
```

```python
# we can change the estimator function
sns.barplot(data = tips, x = 'sex', y = 'tip', estimator=np.min);
```

```python
sns.barplot(data = tips, x = 'sex', y = 'tip', estimator=np.max);
```

```python
sns.barplot(data = tips, x = 'sex', y = 'tip', estimator=np.mean);
```

```python
sns.barplot(data = tips, x = 'sex', y = 'tip', estimator=np.sum);
```

```python
fig, ax = plt.subplots(nrows=1, ncols=2, figsize = (10, 5))
sns.countplot(x = 'day', data = tips, ax = ax[0])
sns.barplot(x = 'day', y = 'total_bill', data = tips, ax = ax[1]);
```

```python
fig, ax = plt.subplots(nrows=1, ncols=2, figsize = (10, 5))
sns.countplot(x = 'day', data = tips, ax = ax[0])
sns.barplot(x = 'day', y = 'total_bill', data = tips, estimator=np.sum, ax = ax[1]);
```
