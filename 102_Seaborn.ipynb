## Distribution Plot

```python
tips.head()
```

```python
# we use distribution plots to understand numerical continuous columns 
sns.displot(x = tips['total_bill']);
```

```python
# we use distribution plots to understand numerical continuous columns 
sns.displot(data = tips, x = 'total_bill');
```

```python
# bins -> number of bars
sns.displot(data = tips, x = 'total_bill', bins = 50);
```

```python

# there are lots of distribution plots, to show the distribution of a numerical data, the general name is distribution plot
# but there are more specific ones as well, 
# histplot --> histogram plot, looks like continuous barchart, connected.
# kdeplot --> looks like a line plot, showing the shape of distribution, and where data is located most.
sns.displot(data = tips, x = 'total_bill', kde = True);
```

```python
# kind -> type of plot, kide of distribution plot
sns.displot(data = tips, x = 'total_bill', kind = 'kde');
```

```python
sns.displot(data = tips, x = 'total_bill', kde = True, bins = 240);
```

```python
fig, ax = plt.subplots(1, 2, figsize = (12, 6))
sns.scatterplot(data = tips,
                x = 'total_bill',
                y = 'tip',
                hue = 'sex',
                s =100, ax = ax[0])
sns.displot(data = tips, x = 'total_bill', kde = True, ax = ax[1]);
# displot doesn't plot side by side
```

```python
fig, ax = plt.subplots(1, 2, figsize = (12, 6))
sns.scatterplot(data = tips,
                x = 'total_bill',
                y = 'tip',
                hue = 'sex',
                s =100, ax = ax[0])
sns.histplot(data = tips, x = 'total_bill', kde = True, ax = ax[1]);
```

## histplot

```python
sns.histplot(data = tips, x = 'total_bill', kde = True, bins = 60);
```

```python
fig, ax = plt.subplots(1, 2, figsize = (12, 6))
sns.rugplot(data = tips, x = 'total_bill', ax = ax[0])
sns.histplot(data = tips, x = 'total_bill', kde = True, ax = ax[1]);
```

```python
fig, ax = plt.subplots(1, 2, figsize = (12, 6))
sns.rugplot(data = tips, x = 'total_bill', height = 0.8,  ax = ax[0]) # hight -> length of lines
sns.histplot(data = tips, x = 'total_bill', kde = True, ax = ax[1]);
```

```python
sns.displot(data = tips, x = 'total_bill', kde = True, rug = True);
# rug -> add rug plot to distribution plot
```

## 2D KDE

```python
tips.head(2)
```

```python
sns.kdeplot(data = tips, x = 'total_bill', y = 'tip');
# the presence of outliers based on far away and separated skirts.
# correlation between two numeric variables
# if too dense, or the lines are too close, means too much data near that point, 
# if less dense, means less data present there, and if they are far from each other, means the data points are separated.
# skewness
```

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip');
# the presence of outliers based on far away and separated skirts.
# correlation between two numeric variables
# if too dense, or the lines are too close, means too much data near that point, 
# if less dense, means less data present there, and if they are far from each other, means the data points are separated.
# skewness
```

```python
sns.kdeplot(data = tips, x = 'total_bill', y = 'tip', color = 'red');
```

```python
sns.kdeplot(data = tips, x = 'total_bill', y = 'tip', color = 'green', shade = True);
```

```python
fig, ax = plt.subplots(1, 2, figsize = (10, 5))
sns.kdeplot(data = tips, x = 'total_bill', y = 'tip', color = 'yellow', shade = True, ax = ax[0])
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', color = 'yellow', ax = ax[1]);
```

```python
fig, ax = plt.subplots(1, 2, figsize = (10, 5))
sns.kdeplot(data = tips, x = 'total_bill', y = 'tip', color = 'yellow', shade = True, ax = ax[1])
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', color = 'yellow', ax = ax[0]);
```

```python
fig, ax = plt.subplots(2, 2, figsize = (10, 7))
sns.kdeplot(data = tips, x = 'total_bill', y = 'tip', color = 'purple', shade = True, ax = ax[0][0])
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', color = 'purple', ax = ax[0][1])
sns.rugplot(data = tips, x = 'total_bill', height = 0.8,  ax = ax[1][0], color = 'purple')
sns.histplot(data = tips, x = 'total_bill', kde = True, ax = ax[1][1], color = 'purple');
```
