# Distribution Plots

## Scatter Plot

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip');
```

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip');
```

## Regplot

```python
sns.regplot(data = tips, x = 'total_bill', y = 'tip');
```

### Hue

- allows you to see different categorical data

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'smoker');
```

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'sex');
```

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue = 'day');
```

### style

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', style='sex');
```

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', style='day');
```

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', style='sex', hue = 'day');
```

```python
# we can give full address of parameters, if we have multiple differents data sets, we can use them in same code
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], style=tips.sex, hue = tips.day);
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], style=tips.sex);
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue = tips.smoker);
```

```python
fig, ax = plt.subplots(1, 2, figsize = (10, 5))
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], style=tips.sex, ax = ax[0])
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue = tips.smoker, ax = ax[1]);
```

### Palette

- different color coding, color combination

```python
# https://seaborn.pydata.org/tutorial/color_palettes.html 
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue=tips.sex, palette='plasma');
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], hue=tips.sex, palette='viridis');
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], style=tips.sex, hue = tips.day, palette='plasma');
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip']);
```

### alpha (density)

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], alpha = 0.5);
```

```python
sns.scatterplot(x = tips['total_bill'], y = tips['tip'], style=tips.sex, hue = tips.day, palette='plasma', alpha = 0.4);
```

### s --> Marker Size

```python
sns.scatterplot(data = tips,
                x = 'total_bill',
                y = 'tip',
                style='sex',
                hue = 'day',
                palette='plasma',
                alpha = 0.4,
                s =100);
```

```python
sns.scatterplot(data = tips,
                x = 'total_bill',
                y = 'tip',
                style='sex',
                hue = 'day',
                palette='cubehelix',
                alpha = 0.7,
                s =250);
```
